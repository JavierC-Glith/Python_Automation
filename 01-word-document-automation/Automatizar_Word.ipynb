```python
import  pandas as pd
from datetime import datetime
from docxtpl import DocxTemplate
```

```python
doc = DocxTemplate("Plantilla.docx")
```

```python
nombre = "Javier Cueto"
telefono = "+569 74965146"
correo = "profesor@gmail.com"
fecha = datetime.today().strftime("%d/%m/%Y")
```

```python
constantes = {'nombre': nombre, "telefono": telefono, "correo": correo, "fecha": fecha}
```

```python
df = pd.read_excel('Alumnos.xlsx')
```

```python
for i,f in df.iterrows():
    contenido = {'nombre_alumno': f["Nombre del Alumno"], 'nota_mat': f["Mat"], 'nota_fis': f["Fis"], 'nota_qui': f["Qui"]}
    contenido.update(constantes)

    doc.render(contenido)
    doc.save(f"Notas_de_{f['Nombre del Alumno']}.docx")
```
